```python
import numpy as np
import matplotlib.pyplot as plt
```

# Angular definition function

```python
# Electron mass in MeV
m_e = 0.51099895


def angular_correlation(Ee, u, v, phi,
                        a=0., b=0., c=0.,
                        A=0., B=0., D=0.,
                        P=1., alignment=0.):
    """
    Angular correlation function for beta decay.

    Ee        : electron/positron total energy [MeV]
    u         : cos(polarization, positron)
    v         : cos(polarization, neutrino)
    phi       : relative azimuthal angle [rad]
    P         : vector polarization magnitude
    alignment : nuclear alignment
    """

    beta = np.sqrt(1. - (m_e / Ee)**2)

    s = np.sqrt((1. - u**2) * (1. - v**2))

    # Angular correlations
    x = u*v + s*np.cos(phi)
    T = s*np.sin(phi)

    # Correlation function
    W = (
        1.
        + b*m_e/Ee
        + a*beta*x
        + A*P*beta*u
        + B*P*v
        + D*P*beta*T
        + c*alignment*beta*(u*v - x/3.)
    )

    return W
```

# Angular minimum calculation

```python
def angular_minimum(Ee, u, v,
                    a=0., b=0., c=0.,
                    A=0., B=0., D=0.,
                    P=1., alignment=0.):

    beta = np.sqrt(1. - (m_e / Ee)**2)

    s = np.sqrt((1. - u**2) * (1. - v**2))

    # Terms independent of phi
    W0 = (
        1.
        + b*m_e/Ee
        + beta*(a + 2.*c*alignment/3.)*u*v
        + A*P*beta*u
        + B*P*v
    )

    # Coefficient of cos(phi)
    C = beta*s*(a - c*alignment/3.)

    # Coefficient of sin(phi)
    S = D*P*beta*s

    return W0 - np.sqrt(C**2 + S**2)
```

# Angular calculation and angular minimum comparison

```python
# Fixed kinematics
Ee = 1.0  # MeV, total positron energy

u = 0.4
v = -0.3

# Example coefficients
params = {
    "a": 0.25,
    "b": -0.05,
    "c": 0.0,
    "A": 0.4,
    "B": -0.2,
    "D": 0.3,
    "P": 1.0,
    "alignment": 0.0
}

# Numerical scan over phi
phi = np.linspace(0., 2.*np.pi, 4001)

W = angular_correlation(Ee, u, v, phi, **params)

W_min_numerical = np.min(W)

# Analytical result
W_min_analytical = angular_minimum(
    Ee, u, v, **params
)

print(f"Numerical minimum : {W_min_numerical:.8f}")
print(f"Analytical minimum: {W_min_analytical:.8f}")
print(f"Difference        : {abs(W_min_numerical-W_min_analytical):.2e}")
```

# Plot comparison

```python
plt.figure(figsize=(8, 5))

plt.plot(
    phi,
    W,
    label=r"$W(\phi)$"
)

plt.axhline(
    W_min_analytical,
    linestyle="--",
    label=r"Analytical minimum"
)

plt.scatter(
    phi[np.argmin(W)],
    W_min_numerical,
    zorder=3,
    label="Numerical minimum"
)

plt.xlabel(r"$\phi$ [rad]")
plt.ylabel(r"$W(\phi)$")

plt.title("Numerical vs Analytical Angular Minimization")

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("numerical_vs_analytical_minimum_comparison.pdf")
plt.show()
```

# Random tests

```python
# Randomized validation of the analytical minimum

rng = np.random.default_rng(2026)

N_tests = 1000
N_phi = 2001

phi_grid = np.linspace(0., 2.*np.pi, N_phi)

differences = []

for i in range(N_tests):

    # Random kinematics
    Ee = rng.uniform(0.6, 2.0)
    u = rng.uniform(-1., 1.)
    v = rng.uniform(-1., 1.)

    # Random correlation coefficients
    params = {
        "a": rng.uniform(-1., 1.),
        "b": rng.uniform(-0.5, 0.5),
        "c": rng.uniform(-1., 1.),
        "A": rng.uniform(-1., 1.),
        "B": rng.uniform(-1., 1.),
        "D": rng.uniform(-1., 1.),
        "P": rng.uniform(0., 1.),
        "alignment": rng.uniform(-1., 1.)
    }

    # Numerical minimum
    W_values = angular_correlation(
        Ee, u, v, phi_grid, **params
    )

    W_num = np.min(W_values)

    # Analytical minimum
    W_ana = angular_minimum(
        Ee, u, v, **params
    )

    differences.append(abs(W_num - W_ana))

differences = np.asarray(differences)

print(f"Tests performed : {N_tests}")
print(f"Maximum difference: {differences.max():.3e}")
print(f"Mean difference   : {differences.mean():.3e}")
```

# 3D Grid Scan

```python
def angular_correlation_cpp_reference(
    Ee, u, v, phi,
    a=0., b=0., c=0.,
    A=0., B=0., D=0.,
    P=1., alignment=0.
):
    beta = np.sqrt(1. - (m_e/Ee)**2)

    theta_e = np.arccos(u)
    theta_nu = np.arccos(v)

    elDir = np.array([
        np.sin(theta_e),
        0.,
        np.cos(theta_e)
    ])

    enuDir = np.array([
        np.sin(theta_nu)*np.cos(phi),
        np.sin(theta_nu)*np.sin(phi),
        np.cos(theta_nu)
    ])

    polDir = np.array([0., 0., 1.])

    # Effective coefficients passed to the C++ function
    A_eff = P*A
    B_eff = P*B
    D_eff = P*D
    c_eff = -alignment*c

    W = (
        1.
        + b*m_e/Ee
        + a*beta*np.dot(elDir, enuDir)
        + A_eff*beta*np.dot(elDir, polDir)
        + B_eff*np.dot(enuDir, polDir)
        + D_eff*beta*np.dot(
            polDir, np.cross(elDir, enuDir)
        )
        + c_eff*beta*(
            np.dot(elDir, enuDir)/3.
            - np.dot(elDir, polDir)
            * np.dot(enuDir, polDir)
        )
    )

    return W
```

```python
def minimum_azimuth(Ee, u, v,
                    a=0., b=0., c=0.,
                    A=0., B=0., D=0.,
                    P=1., alignment=0.):

    beta = np.sqrt(1. - (m_e / Ee)**2)

    s = np.sqrt((1. - u**2) * (1. - v**2))

    C = beta*s*(a - c*alignment/3.)
    S = D*P*beta*s

    phi_min = (np.arctan2(S, C) + np.pi) % (2.*np.pi)

    return phi_min
```

```python
def evaluate_coefficients(Ee, params):

    evaluated = {}

    for name, value in params.items():

        if callable(value):
            evaluated[name] = value(Ee)
        else:
            evaluated[name] = value

    return evaluated
```

```python
def grid_scan(E_min, E_max, params,
              N_E=50, N_u=100, N_v=100):

    energies = np.linspace(E_min, E_max, N_E)
    u_values = np.linspace(-1., 1., N_u)
    v_values = np.linspace(-1., 1., N_v)

    U, V = np.meshgrid(u_values, v_values, indexing="ij")

    global_min = np.inf
    minimum_location = None

    for Ee in energies:

        evaluated_params = evaluate_coefficients(Ee, params)

        W = angular_minimum(
            Ee, U, V, **evaluated_params
        )

        index = np.unravel_index(
            np.argmin(W), W.shape
        )

        if W[index] < global_min:

            global_min = W[index]

            minimum_location = (
                Ee,
                U[index],
                V[index]
            )

    return global_min, minimum_location
```

# 3D Grid Scan: tests

```python
params_zero = {
    "a": 0.,
    "b": 0.,
    "c": 0.,
    "A": 0.,
    "B": 0.,
    "D": 0.,
    "P": 1.,
    "alignment": 0.
}

W_min, location = grid_scan(
    E_min=0.6,
    E_max=2.0,
    params=params_zero
)

print(f"Minimum W: {W_min:.8f}")
print(f"Location (Ee, u, v): {location}")
print(f"Physical validity: {'PASS' if W_min >= 0 else 'FAIL'}")
```

```python
params_invalid = params_zero.copy()

params_invalid["A"] = 2.0

W_min, location = grid_scan(
    E_min=0.6,
    E_max=2.0,
    params=params_invalid
)

print(f"Minimum W: {W_min:.8f}")
print(f"Location (Ee, u, v): {location}")
print(f"Physical validity: {'PASS' if W_min >= 0 else 'FAIL'}")
```

```python
params_combined = {
    "a": 0.25,
    "b": -0.05,
    "c": 0.0,
    "A": 0.4,
    "B": -0.2,
    "D": 0.3,
    "P": 1.0,
    "alignment": 0.0
}

W_min, location = grid_scan(
    E_min=0.6,
    E_max=2.0,
    params=params_combined,
    N_E=50,
    N_u=100,
    N_v=100
)

print(f"Minimum W: {W_min:.8f}")
print(f"Location (Ee, u, v): {location}")
print(f"Grid result: {'NO NEGATIVE POINTS FOUND' if W_min >= 0 else 'INVALID'}")
```

```python
E_max = 2.0

params_energy = {
    "a": lambda Ee: 0.2 + 0.1*Ee/E_max,
    "b": -0.05,
    "c": 0.0,
    "A": 0.4,
    "B": -0.2,
    "D": lambda Ee: 0.1 + 0.2*Ee/E_max,
    "P": 1.0,
    "alignment": 0.0
}

W_min, location = grid_scan(
    E_min=0.6,
    E_max=E_max,
    params=params_energy,
    N_E=50,
    N_u=100,
    N_v=100
)

print(f"Minimum W: {W_min:.8f}")
print(f"Location (Ee, u, v): {location}")
print(
    "Grid result:",
    "NO NEGATIVE POINTS FOUND" if W_min >= 0 else "INVALID"
)
```

```python
rng = np.random.default_rng(2026)

N_tests = 1000
differences = []

for _ in range(N_tests):

    Ee = rng.uniform(0.6, 2.0)
    u = rng.uniform(-1., 1.)
    v = rng.uniform(-1., 1.)
    phi = rng.uniform(0., 2.*np.pi)

    params = {
        "a": rng.uniform(-1., 1.),
        "b": rng.uniform(-0.5, 0.5),
        "c": rng.uniform(-1., 1.),
        "A": rng.uniform(-1., 1.),
        "B": rng.uniform(-1., 1.),
        "D": rng.uniform(-1., 1.),
        "P": rng.uniform(0., 1.),
        "alignment": rng.uniform(-1., 1.)
    }

    W_python = angular_correlation(
        Ee, u, v, phi, **params
    )

    W_cpp_reference = angular_correlation_cpp_reference(
        Ee, u, v, phi, **params
    )

    differences.append(abs(W_python - W_cpp_reference))

differences = np.asarray(differences)

print(f"Maximum difference: {differences.max():.3e}")
print(f"Mean difference: {differences.mean():.3e}")
```
